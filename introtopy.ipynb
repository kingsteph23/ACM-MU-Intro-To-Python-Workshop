```python
#iterations
for i in range(1,6): # iterating through each element within  the range (1,6]. However 6 isnt includes. so technically printing from 1-5
  print(i) #print each indiviual element with that range(1-5)

print()#making a space

for i in range(1,6): ## iterating through each element within  the range (1,6]. However 6 isnt includes. so technically printing from 1-5
  print(i)# printing each indivdual challenge
  if i == 3:# if the element(i) == 3, break(leave) the loop
    break# leave the stop
print()#making a space
while i < 6:# while loop. same concept as first loop. Go up to 6, butd dont include 6
  print(i)# printi indivdual elements
  i+=1# increment by 1
```

```python
#functions

def add(x,y): #creating a function that does addition. x and y represents the parametrs that will be used
  print(x+y) #printing the sum

def subtract(x,y):#creating a function that does subtraction. x and y represents the parametrs that will be used
  print(x-y)# printing the difference

def multiply(x,y): #creating a function that does multiplication. x and y represents the parametrs that will be used
  product = x*y# creating a variable called producted and assigned it to the product of x,y
  return product #returning the specifed value
add(2,3)#passing in arguements
subtract(3,2)#passing in arguments for subtraction
multiply(6,7)# passing in arguements for multiplication
```

```python
#lists
jpn = ["honda","toyota","nissan","lexus","mazda"] # listing of strings(consisted of japanese car brands)
usa = ["ford","chevy","buick","dodge","ram"] #list of strings(consist of american car brands)

print(jpn)#printing full  japan list
print(usa)# printing full usa list
print(jpn[1])# printing the 2nd value present in the list(note: the index of the list begin at 0,1.... So the first val is at jpn[0],2nd jpn[1]....)
print(jpn[2:])#printing values starting from the 2 index onward
print(jpn[:2])#printing all the values prior to the 2nd index but not including the 2 index
print(jpn[-1])#printing the last value at the last index
print(usa[0])# printing first value present at first index of usa list

jpn.sort() #sorting list in ascending order(a,b,c,d.... or 1,2,3... if working with numbers)
print(jpn)# printing sorted list
jpn.sort(reverse=True)#sorting list in reverse order(z,y,x... or 10,9,8... if working with numbers)
print(jpn)#printing decended order
jpn.pop(1)# remove the value that is present at index 1
print(jpn)#print list with the index removed
jpn.insert(1,"suzuki")# insert suzuki at index 1
print(jpn)# print japan list with suzuki
jpn.append("lexus")#add lexus to end of jpn array
print(jpn)#print japan with newly added lexus
jpn.remove("lexus")# remove the first presence of lexus with the list
print(jpn)#print jpn with first presence of lexus removed
#jpn.clear()#Make list empty. Remove all the values
usa.append(jpn)# append the japanese list to the american list. Add the japanese list to the american list
print(usa)# print combined american and japanese car list
```
